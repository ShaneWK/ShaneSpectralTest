```python
# LCG Lattice Visualization Notebook
### Interactive geometric visualization of k-dimensional LCG lattices

# Purpose of This Notebook

## This notebook provides an *interactive geometric visualization* of the lattice structure,
## which are generated by Linear Congruential Generators (LCGs). 
## The spectral test evaluates the spacing between hyperplanes in these lattices, 
## this notebook helps illustrate those geometric patterns directly.

# The goals of this notebook are:

## - To visualize k-dimensional LCG tuples in 3D.
## - To demonstrate the difference between good and bad LCGs.
## - To provide an interactive widget for exploring different generators and parameters.
## - To support presentation slides with clear geometric screenshots.

# What You Can Do Here

## Use the interactive widget below to:

## - Select an LCG preset (e.g., Park–Miller, Knuth MMIX, RANDU).
## - Adjust the number of generated points.
## - Rotate and zoom the 3D lattice plot.
## - Compare good vs. bad generators visually.

# Interpretation

## - **Good LCGs** produce dense, uniform point clouds with no visible planes.
## - **Bad LCGs** (like RANDU) produce obvious parallel planes or degeneracies.
## - These geometric patterns correspond directly to the hyperplane spacing measured by the spectral test.

# This notebook is intended for demonstration and presentation purposes. 
## It complements the full spectral test implementation by showing the underlying geometry that the test quantifies.


import numpy as np
import plotly.graph_objects as go
from ipywidgets import interact, IntSlider, Dropdown

def lcg(a, c, m, seed, n):
    x = seed
    out = []
    for _ in range(n):
        x = (a * x + c) % m
        out.append(x / m)
    return np.array(out)

def make_tuples(seq, k):
    return np.column_stack([seq[i:len(seq)-k+i+1] for i in range(k)])

def plot_lattice(points, title="LCG Lattice (3D)"):
    fig = go.Figure(data=[go.Scatter3d(
        x=points[:,0],
        y=points[:,1],
        z=points[:,2],
        mode='markers',
        marker=dict(size=3, opacity=0.7)
    )])
    fig.update_layout(
        title=title,
        scene=dict(
            xaxis_title='x₁',
            yaxis_title='x₂',
            zaxis_title='x₃'
        ),
        width=800,
        height=700
    )
    fig.show()

LCG_PRESETS = {
    "Knuth MMIX (Good)": {"a": 6364136223846793005, "c": 1442695040888963407, "m": 2**64, "seed": 1},
    "Park–Miller (Good)": {"a": 16807, "c": 0, "m": 2**31 - 1, "seed": 1},
    "RANDU (Infamously Bad)": {"a": 65539, "c": 0, "m": 2**31, "seed": 1},
    "Bad LCG Example": {"a": 5, "c": 1, "m": 2**16, "seed": 7}
}

def visualize_lcg(preset, n, k):
    params = LCG_PRESETS[preset]
    seq = lcg(params["a"], params["c"], params["m"], params["seed"], n)
    pts = make_tuples(seq, k)

    if k == 3:
        plot_lattice(pts, title=f"{preset} — 3D Lattice (n={n})")
    else:
        print("Only k=3 is visualized in 3D. Try k=3.")

interact(
    visualize_lcg,
    preset=Dropdown(options=list(LCG_PRESETS.keys()), value="Park–Miller (Good)"),
    n=IntSlider(value=5000, min=1000, max=20000, step=1000),
    k=IntSlider(value=3, min=2, max=3)
)
```
